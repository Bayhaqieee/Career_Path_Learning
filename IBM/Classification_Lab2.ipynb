## Configure Environment

```python
from langchain_community.llms import Replicate
import os
from google.colab import userdata

# Set the API token
api_token = userdata.get('REPLICATE_API_TOKEN')
os.environ["REPLICATE_API_TOKEN"] = api_token

# Model setup
model = "ibm-granite/granite-3.2-8b-instruct"
output = Replicate(
  model=model,
  replicate_api_token=api_token,
)
```

## Test initial classification output with default values for prompt parameters

```python
# Define the customer reviews
customer_reviews = [ "batrainya jelek!."]

# Refine the prompt to include reviews
reviews_text = "\n".join([f"Review {i+1}: {review}" for i,
review in enumerate(customer_reviews)])
```

```python
reviews_text
```

**Remember**: Adjusting parameters that control token selection, length, or diversity can significantly refine the quality of the responses. These include:
- top_k: Limits the number of possible next tokens considered when generating text, selecting from the "top k" most likely options based on their probability. This focuses the output and reduces the likelihood of unexpected or irrelevant words, acting as a filter for the most probable next words.
- top_p: Influences diversity by restricting token selection to the most probable options within a cumulative probability threshold. It ensures that the model chooses from a more focused set of options, improving response coherence while maintaining diversity.
- max_tokens: Controls the maximum length of the model’s response. By setting a token limit, this parameter prevents overly lengthy responses while helping to focus the model’s output within a manageable size.
- min_tokens: Sets a minimum length for the output, ensuring the model generates sufficiently detailed responses, which is useful for tasks requiring comprehensive explanations or summaries
- random_seed: Ensures reproducibility by fixing the randomness in model output
- repetition_penalty: Reduces redundancy by penalizing repeated tokens in the output. This encourages the model to generate more diverse and varied language, preventing it from repeating the same words or phrases unnecessarily.
- stopping_criteria: Defines the conditions under which the model stops generating tokens
- stopping_sequence: Halts the response when the model encounters a predefined string, such as a newline character or specific word

```python
# Set model parameters for prompting with default values
parameters = {
"top_k": 0,
"top_p": 1.0,
"max_tokens": 256,
"min_tokens": 0,
"random_seed": 1,
"repetition_penalty": 1.0,
"stopping_criteria": "length (256 tokens)",
"stopping_sequence": None
}
```

```python
# Add initial prompt
refined_prompt = f""":
Classify these reviews as positive, negative, or mixed, and tag relevant focus areas such as battery life, screen quality, or performance
{reviews_text} """

# Invoke the model
response = output.invoke(refined_prompt, parameters=parameters)

# Print the response print("Granite Model Refined Response:\n")
print(response)
```

Note that the output highlights the focus areas effectively, addressing positive and negative aspects. However, the response is not concise, making it less suitable for quick evaluation. To address this, we need to refine the prompt by adjusting the top_k parameter to ensure more granular and concise sentiment categorization.
